```python
columns_to_check = [
    "OccupationType",
    "Healthy",
    "Habits",
    "Diabetes2",
    "BloodPressure",
    "Cholesterol",
    "Overweight",
    "AgeGrp",
    "scnZone",
    "healthRisk",
    "DiabetesFoot"
]
for col in columns_to_check:
    print("\n" + "="*50)
    print(col)
    print("="*50)
    print(df[col].value_counts(dropna=False))
```

```python
numeric_columns = ["Weight", "Height", "BMI", "Measurement.Att.Date"]
for col in numeric_columns:
    print("\n" + "="*50)
    print(col)
    print("="*50)
    print("Missing:", df[col].isnull().sum())
    print("Unique values:", df[col].nunique())
    if col != "Measurement.Att.Date":
        print(df[col].describe())
    else:
        print("Minimum:", df[col].min())
        print("Maximum:", df[col].max())
        print("First 20 values:")
        print(df[col].head(20).tolist())
```

```python
# Duplicates
df.duplicated().sum()
```

```python
# Outliers
numeric_cols = ["Weight", "Height", "BMI"]
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print("\n" + "="*50)
    print(col)
    print("="*50)
    print("Lower bound:", lower)
    print("Upper bound:", upper)
    print("Number of outliers:", len(outliers))
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
```

```python
# Checking actual Outlier Records
for col in ["Weight", "Height", "BMI"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    print("\n", col, "OUTLIERS")
    print(df[(df[col] < lower) | (df[col] > upper)][[col]].head(20))
```

```python
# Check wheather BMI is consistent with Height and Weight
df["BMI_calculated"] = (
    df["Weight"] / (df["Height"] ** 2)
).round(1)
df["BMI_difference"] = (
    df["BMI"] - df["BMI_calculated"]
).abs()
df[["Weight", "Height", "BMI", "BMI_calculated", "BMI_difference"]].head(10)
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Tells us Affected rows
df[df[["Weight", "Height", "BMI"]].isnull().any(axis=1)][
    ["Weight", "Height", "BMI"]
].to_string()
```

```python
# Filling the missing values
df["Weight"] = df["Weight"].fillna(df["Weight"].median())
df["Height"] = df["Height"].fillna(df["Height"].median())
df["BMI"] = df["BMI"].fillna(df["BMI"].median())
```

```python
# Filling Categorical Missing Values
categorical_cols = [
    "DiabetesFoot",
    "Diabetes2",
    "BloodPressure",
    "Cholesterol",
    "OccupationType",
    "StaffEducation",
    "Health",
    "healthRisk"
]
for col in categorical_cols:
    df[col] = df[col].fillna("Not Recorded")
```

```python
# Rechecking all missing values
print("Remaining missing values:")
print(df.isnull().sum())
```

```python
df.drop(
    columns=["BMI_calculated", "BMI_difference"],
    inplace=True
)
print(df.columns.tolist())
```
